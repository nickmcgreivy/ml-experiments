```python
bucket_name = 'nickm-ml-datasets'
prefix = 'imagenet'
local_root = os.getcwd()
!aws s3 sync s3://{bucket_name}/{prefix}/meta {local_root}/meta
```

```python
val_dataset = load_imagenet(bucket_name, prefix, split='val')
```

```python
train_dataset = load_imagenet(bucket_name, prefix, split='train')
```

```python
val_dl = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=2)
```

```python
t1 = time()
for i, (img, label) in enumerate(val_dl):
    if i == 0:
        print(img.shape, label.shape)
    if i == 9:
        break
t2 = time()
print(f'{(t2 - t1)/10:.2f} seconds per batch')
```

```python
train_dl = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=2)
```
